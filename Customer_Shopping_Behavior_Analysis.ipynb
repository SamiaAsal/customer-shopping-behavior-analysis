**Data Quality**

```python
#check for missing values
data.isnull().sum()
```

```python
# check for Unnmamed columns
print(data[['Unnamed: 19', 'Unnamed: 20']].head(10).to_string())
```

```python
# drop the Unnamed columns
data = data.drop(columns = ['Unnamed: 19', 'Unnamed: 20'])
```

```python
print('Data Duplicates:',data.duplicated().sum())
print('Data Types:',data.dtypes)
```

```python
data['Purchase Amount (USD)'] = (data['Purchase Amount (USD)'].astype(str).str.replace('$', ''))
data['Purchase Amount (USD)'] = pd.to_numeric(data['Purchase Amount (USD)'])
```

```python
# check the data type of the 'Purchase Amount (USD)' column
(data['Purchase Amount (USD)'].dtype)
```

The dataset contained 3900 rows and 21 columns.


1.   Missing Values: Two unnamed columns containg missing values and were removerd after checking their contents. the remaining columns had no missing values.
2.   Data types: The Purchase Amount column was stored as a string because the values contained $. I removed the dollar signs and convered the column to a numeric data type(Float)
3. Duplicated Records: I Found no duplicates.

**Descriptive Statistics**

```python
col_num = ['Age', 'Purchase Amount (USD)', 'Review Rating', 'Previous Purchases']

for col in col_num:
    print(f' Number {col}: {data[col].count()}')
    print(f' Sum {col}: {data[col].sum()}')
    print(f' Mean {col}: {data[col].mean().round(2)}')
    print(f' Min {col}: {data[col].min()}')
    print(f' Max {col}: {data[col].max()}')
    print()
```


**Key-findings**



*   The dataset contain 3,900 records. the average customer age is 44.07 years, with age ranging from 18 to 70 years.
*   The total purchase amount is 233081$, and average is 59.76
*   The average view rating is 3.75 out of 5. indicate the average rating is above the midpoint of the scale.
*   The average number of previous purchases is 25.35 with values raning from 1 to 50.

```python
clothing = data[data['Category'] == 'Clothing']
clothing_sales = clothing.groupby('Item Purchased')['Purchase Amount (USD)'].sum().sort_values(ascending=False)
clothing_sales
```

```python
Blouse = clothing[clothing['Item Purchased'] == 'Blouse']
Blouse_size = Blouse.groupby('Size')['Purchase Amount (USD)'].sum().sort_values(ascending=False)
Blouse_size
```
